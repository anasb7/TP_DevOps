# Demo: reading a mounted dataset

`/home/jovyan/datasets` is bind-mounted read-only from the host `datasets/` folder.

This notebook is read-only too: use *File → Save As…* into `work/` to keep your changes.

```python
import pandas as pd

DATASET = "/home/jovyan/datasets/trees.csv"
trees = pd.read_csv(DATASET)
trees.head()
```

```python
trees.describe()
```

## Average height per species

```python
by_species = trees.groupby("species")["height_m"].mean().sort_values()
by_species.plot.barh(color="#2e9e44", title="Average height (m)");
```

Write a file into `work/`, restart the containers, and check it is still there.
